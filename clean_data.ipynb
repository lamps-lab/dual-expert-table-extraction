## Cleaning the data

It iterates through all `.txt` files in `./input/train/nougat`, and for each file it deletes the file if it is empty or contains `[MISSING_PAGE_POST]`. When it deletes a bad text file, it also removes the corresponding image and structure file in the training folders.

```python
from pathlib import Path
import random
import shutil
import json
import re
```

```python
nougat_list = list(Path("./input/test/nougat").glob("*.txt"))
```

```python
print(f"Total nougat tables: {len(nougat_list)}")
```

```python
nougat_stems = {p.stem for p in nougat_list}

img_folder = Path("./input/test/structure_processed")

for img_path in img_folder.iterdir():
    if img_path.is_file() and img_path.stem not in nougat_stems:
        print(f"Deleting: {img_path}")
        img_path.unlink()
```

```python
def is_valid_table(text: str) -> bool:
    """
    Check whether the txt file contains a properly formatted LaTeX table.
    """

    # Must contain both begin/end tabular
    if r"\begin{tabular}" not in text or r"\end{tabular}" not in text:
        return False

    # Count should match
    if text.count(r"\begin{tabular}") != text.count(r"\end{tabular}"):
        return False

    # Basic regex validation
    pattern = re.compile(
        r"\\begin\{tabular\}.*?\\end\{tabular\}",
        re.DOTALL
    )

    return bool(pattern.search(text))

NQ_DIR = Path("./input/test/nougat") 
txt_files = list(NQ_DIR.glob("*.txt"))

deleted = 0
kept = 0

for txt_file in txt_files:

    try:
        content = txt_file.read_text(encoding="utf-8").strip()

        # Remove empty files
        if not content:
            img_folder = Path(f"./input/test/img/{txt_file.stem}.png")
            struct_folder = Path(f"./input/test/structure_processed/{txt_file.stem}.json")

            img_folder.unlink(missing_ok=True)
            struct_folder.unlink(missing_ok=True)
            txt_file.unlink()
            deleted += 1
            continue

        # Remove invalid formatting
        if not is_valid_table(content):
            img_folder = Path(f"./input/test/img/{txt_file.stem}.png")
            struct_folder = Path(f"./input/test/structure_processed/{txt_file.stem}.json")

            img_folder.unlink(missing_ok=True)
            struct_folder.unlink(missing_ok=True)
            txt_file.unlink()
            deleted += 1
            continue

        kept += 1

    except Exception as e:
        print(f"Error processing {txt_file.name}: {e}")

print(f"Kept: {kept}")
print(f"Deleted: {deleted}")
print("Finished cleaning dataset.")
```

```python
folder = Path("./input/train/nougat")

for txt_path in folder.glob("*.txt"):
    if txt_path.is_file():
        with open(txt_path, "r") as f:
            content = f.read()
            if content.strip() == "":
                print(f"Deleting empty file: {txt_path}")
                txt_path.unlink()
            elif '[MISSING_PAGE_POST]' in content:
                print(f"Deleting file with [MISSING_PAGE_POST]: {txt_path}")

                img_folder = Path(f"./input/train/img/{txt_path.stem}.png")
                struct_folder = Path(f"./input/train/structure_processed/{txt_path.stem}.png")

                img_folder.unlink(missing_ok=True)
                struct_folder.unlink(missing_ok=True)
                txt_path.unlink()
```
